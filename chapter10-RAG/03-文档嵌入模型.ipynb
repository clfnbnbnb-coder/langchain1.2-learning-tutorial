# 文档嵌入模型

## 1、嵌入模型的选型和初始化

### 情况1：使用智谱平台API

```python
import os
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings

load_dotenv(verbose=True)


embedding_model_zhipu = init_embeddings(
    model="openai:embedding-3",
    api_key=os.getenv("ZHIPUAI_API_KEY"),
    base_url=os.getenv("ZHIPUAI_BASE_URL"),
)
```

### 情况2：使用硅基流动平台API

```python
import os
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings

load_dotenv(verbose=True)


embedding_model_guiji = init_embeddings(
    model="openai:Pro/BAAI/bge-m3",
    api_key=os.getenv("SILICONFLOW_API_KEY"),
    base_url=os.getenv("SILICONFLOW_BASE_URL"),
)
```

或者

```python
from langchain_openai import OpenAIEmbeddings

embedding_model_guiji1 = OpenAIEmbeddings(
    model="Pro/BAAI/bge-m3",
    api_key=os.getenv("SILICONFLOW_API_KEY"),
    base_url=os.getenv("SILICONFLOW_BASE_URL"),
)
```

## 2、句子的向量化

```python

text = "你好，很高兴认识你"

embed_docs = embedding_model_zhipu.embed_query(text)

print(len(embed_docs))

print(embed_docs[:5])
```

```python

text = "你好，很高兴认识你"

embed_docs = embedding_model_guiji1.embed_query(text)

print(len(embed_docs))

print(embed_docs[:5])
```

## 3、文档的向量化

### 举例1：

```python
# 待嵌入的文本列表
texts = [
    "Hi there!",
    "Oh, hello!",
    "What's your name?",
    "My friends call me World",
    "Hello World!"
]
# 生成嵌入向量
embedded_docs = embedding_model_zhipu.embed_documents(texts)
for i in range(len(texts)):
    print(f"{texts[i]}:{embedded_docs[i][:3]}", end="\n\n")
```

### 举例2：

```python
from langchain_community.document_loaders import CSVLoader

# 情况1：加载CSV文件并向量化处理
loader = CSVLoader("../asset/load/02-load.csv", encoding="utf-8")
docs = loader.load_and_split()

# print(len(docs))

# 存放的是每一个chunk的embedding
texts = [doc.page_content for doc in docs]
embeded_docs = embedding_model_zhipu.embed_documents(texts)

print(len(embeded_docs))

for i in range(len(texts)):
    print(f"{texts[i]}:\n{embeded_docs[i][:3]}", end="\n\n")
```
